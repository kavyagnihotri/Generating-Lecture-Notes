```python
import pandas as pd
import csv
```

```python
def URI_and_their_entities(entity_file):
    mp = {}
    df = pd.read_csv(entity_file)
    for index, row in df.iterrows():
        entity_URI = str(row['entity_URI']).strip()
        entity_name = str(row['entity_name']).strip()
        if entity_URI in mp:
            mp[entity_URI].append(entity_name)
        else:
            mp[entity_URI] = [entity_name]
    return mp
```

```python
def unique_entities(unique_entity_file):
    mp = []
    df = pd.read_csv(unique_entity_file)
    for index, row in df.iterrows():
        mp.append(str(row['entity_name']).strip())
    return list(mp)
```

```python
def entity_with_URI(entity_file):
    mp = {}
    df = pd.read_csv(entity_file)
    for index, row in df.iterrows():
        mp[str(row['entity_name']).strip()] = str(row['entity_URI']).strip()
    return mp
```

```python
# make a new csv file such that the first column is the entity name and the following columns are the first_hop_entities
def get_map(map_file, unique_entity_file, entity_file, output_file):
    same_entities = URI_and_their_entities(entity_file)
    unique_entity_names = unique_entities(unique_entity_file)
    entities_URI = entity_with_URI(entity_file)

    # open output_file and map_file using with open
    with open(output_file, 'a') as f, open(map_file, 'r') as f1:
        count = 0
        for line in f1:
            new_row = []
            entity_name = line.split(',')[0].strip()  

            if entity_name not in unique_entity_names:
                continue
            new_row.append(entity_name)
            
            first_hop_entities = line.split(',')[1:]
            for entity in first_hop_entities:
                entity = entity.strip()
                if entity in unique_entity_names:
                    new_row.append(entity)
                else:
                    entity_URI = entities_URI.get(entity)
                    if entity_URI:
                        same_entities_list = same_entities.get(entity_URI)
                        if same_entities_list:
                            for same_entity in same_entities_list:
                                same_entity = same_entity.strip() 
                                if same_entity in unique_entity_names:
                                    new_row.append(same_entity)
                                    break

            f.write(','.join(new_row) + '\n')
            count += 1
            print(count, entity_name)
```

```python
def get_map(map_file, unique_entity_file, entity_file, output_file):
    same_entities = URI_and_their_entities(entity_file)
    unique_entity_names = unique_entities(unique_entity_file)
    entities_URI = entity_with_URI(entity_file)

    # Create an empty list to store rows (dictionaries)
    rows = []

    max_columns = 0  # To track the maximum number of columns encountered

    with open(map_file, 'r') as f1:
        count = 0
        csv_reader = csv.reader(f1)
        for row in csv_reader:
            if not row:
                continue  # Skip empty lines
            
            entity_name = row[0].strip()

            if entity_name not in unique_entity_names:
                continue
            
            new_row = {"entity_name": entity_name}
            
            for entity in row[1:]:
                entity = entity.strip()
                if entity in unique_entity_names:
                    new_row[f"entity_{len(new_row)}"] = entity
                else:
                    entity_URI = entities_URI.get(entity)
                    if entity_URI:
                        same_entities_list = same_entities.get(entity_URI)
                        if same_entities_list:
                            for same_entity in same_entities_list:
                                same_entity = same_entity.strip() 
                                if same_entity in unique_entity_names:
                                    new_row[f"entity_{len(new_row)}"] = same_entity
                                    break
            
            rows.append(new_row)
            
            # Update the maximum number of columns if needed
            max_columns = max(max_columns, len(new_row))

            count += 1
            print(count, entity_name)

    # Generate column names based on the maximum number of columns
    column_names = ["entity_name"] + [f"entity_{i+1}" for i in range(max_columns - 1)]

    # Create the DataFrame with the determined column names
    result_df = pd.DataFrame(rows, columns=column_names)

    # Save the list of dictionaries to a CSV file
    with open(output_file, 'w', newline='') as csvfile:
        fieldnames = result_df.columns
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        
        writer.writeheader()
        for row in rows:
            writer.writerow(row)

```

```python
get_map(map_file, unique_entity_file, entity_file, output_file)
```
